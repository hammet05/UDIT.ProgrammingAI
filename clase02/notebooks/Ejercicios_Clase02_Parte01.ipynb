```python
# Definimos la cadena
texto = "programacion"

# 1. Las 4 primeras letras
print(texto[0:4])

# 2. Las 3 últimas letras
print(texto[-3:])

# 3. Cadena invertida
print(texto[::-1])

# 4. Cada 2 caracteres desde el índice 1
print(texto[1::2])
```

```python
                                                            # Estructuras de control

# 1. Escribe un programa que pida un número al usuario y diga si es positivo, negativo o cero. 

numero = float(input("Introduce un número"))

if numero>0:
    print("El número introducido es positivo")
elif numero<0:
    print("El número introducido es negativo")
else:
    print("El número introducido fue cero")          
        
```

```python
#2. Recorre una lista de nombres e imprime solo aquellos que tengan más de 4 letras.

nombres =['Hamilton',"Oscar","Charles","Kimi","Alex","Max"]

for nombre in nombres:
    if len(nombre) > 4:
        print(nombre)
    continue
```

```python
# 3. Implementa un `while` que vaya contando desde 1 hasta que el usuario escriba `"stop"`.  

contador = 0

while True:
        
    respuesta = input("Escribe stop para detener la secuencia de números")
    
    if respuesta.lower()=="stop":
        break
           
    print(contador)
    contador += 1        
    
    print("Secuencia finalizada")
    
```

```python
# 4. Escribe un bucle que recorra una lista de enteros y construya otra lista con solo los valores pares.
import random

numeros = []
pares=[]

for i in range(30):
    num = random.randint(1,100)
    numeros.append(num)

    if num%2==0:
        pares.append(num)

print(numeros)
print(pares)
```
