```python
# 2. Crea una función `media` que acepte cualquier cantidad de números (`*args`) y calcule la media.  
import random

def generar_numeros(cantidad:int)->list[int]:
    numeros = []
    for i in range(cantidad):
        num=random.randint(1,100)
        numeros.append(num)
    return numeros


def media(*args: float)->float:
    if len(args)==0:
        return 0.0
        
    return sum(args) / len(args)
    

numeros = generar_numeros(10)
print("Numeros: ", numeros)
print("Media: ", media(*numeros))
```

```python
# 3 Usa `map` con una función `lambda` para convertir una lista de temperaturas en grados Celsius a Fahrenheit.  
import random

def generar_lista_temperaturas_celsius(cantidad:int)->list[int]:
    numeros = []
    for i in range(cantidad):
        num=random.randint(1,100)
        numeros.append(num)
    return numeros
    
print("Lista temperaturas Celsius:",generar_lista_temperaturas_celsius(10))

temperaturas_celsius = generar_lista_temperaturas_celsius(15)

temperaturas_fahrenheit = list(map(lambda celsius:(celsius*9/5)+ 32,temperaturas_celsius))
print("Conversion a grados Fahrenheit",temperaturas_fahrenheit)
```

```python
# 7. Dada una lista de palabras, usa **`map`** para pasarlas a *title case* y **`filter`** para quedarte solo con aquellas de longitud ≥ 5.

palabras = ["byte","celular","coche","avión","modem","bicicleta","mouse","carta","oro"]

titles=list(map(lambda palabra: palabra.title(),palabras))

filtro = list(filter(lambda longitud5 : len(longitud5)>5,titles))
print(titles)
print("Palabras con filtro de longitud > 5", filtro)
```

```python
#3 . Implementa una función `concatena(a: str, b: Optional[str]) -> str` que devuelva la concatenación de `a` y `b`, usando cadena vacía si `b` es `None`.

from typing import Optional


def concatena(a: str, b: Optional[str]) -> str:
    if b is None:
        return ''
    else:
        return a+b
    
print(concatena("video ","None"))
    

```

```python
#5. Declara una función `procesar(valor: Union[int, str]) -> str` que:  
#  - Si recibe un entero, devuelva su cuadrado como cadena.  
# - Si recibe un string, lo devuelva en mayúsculas.

from typing import Union


def procesar(valor: Union[int, str]) -> str:
    if isinstance(valor,int):
        return str(valor*valor)
    else: 
        return valor.upper()
    

print(procesar("es un 5"))
```

```python
# Manejo de ficheros
"""
1. Escribe una función que abra un fichero cuyo nombre se pasa como argumento:  
- Si **no existe**, capture la excepción y muestre un mensaje.  
- Si existe, devuelva el número de líneas que contiene.
"""
def cuenta_lineas(nombre):
    try:
        with open(nombre) as fichero:
            return len(fichero.readlines())
    except FileNotFoundError:
        print("El fichero solicitado no existe")
        
#print(cuenta_lineas("datos.txt"))
print(cuenta_lineas("data.txt"))
```

```python
# COMPREHENSION
#1. Crea un set comprehension con los números entre 0 y 20 que sean múltiplos de 3..

multiplos_de_3 = {n for n in range(0, 21) if n % 3 == 0}

print(multiplos_de_3)  # {0, 3, 6, 9, 12, 15, 18}


#2. A partir de la lista `frases = ["hola mundo", "python es genial", "list comprehensions"]`,
#crea una lista con todas las palabras en mayúsculas, usando una comprensión anidada..
frases = ["hello world", "cos',", "you know,","every single program","must have one"]

palabras = [palabra.upper() for frase in frases for palabra in frase.split()]

print(palabras)
```
