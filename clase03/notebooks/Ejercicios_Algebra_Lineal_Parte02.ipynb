```python
import numpy as np

# **Ejemplo práctico**

# Tenemos el siguiente array con temperaturas:


# temps = np.array([15, -3, 22, 31, 29, -5, 35])


# y queremos:
# 1. Filtrar las que superan 30 grados.
# 2. Reemplazar por 0 las que son negativas.
# 3. Clasificarlas como `"Calor"` o `"Fresco"` usando `where`.
temps = np.array([15, -3, 22, 31, 29, -5, 35])

# Filtrar las que superan 30 grados.
temps_mayor_30 = temps[temps>30]

# 2. Reemplazar por 0 las que son negativas.
temps_negativas = temps.copy()
temps_negativas[temps_negativas<0]=0

# 3. Clasificarlas como `"Calor"` o `"Fresco"` usando `where`.
temps_clasificado = np.where(temps>25,"Calor", "Fresco")

print("Temperaturas originales \n",temps)
print("Temperaturas mayores a 30 grados celsius \n",temps_mayor_30)
print("Temperaturas negativas \n",temps_negativas)
print("Temperaturas clasificadas\n",temps_clasificado)
```

```python

# 1. Crea una matriz de 5x5 con los valores del 0 al 24.
import random

def crear_matriz(filas, columnas, minimo, maximo):
    matriz = []
    for i in range(filas):
        fila = []
        for j in range(columnas):
            numero = random.randint(minimo, maximo)
            fila.append(numero)
        matriz.append(fila)
    return matriz


# Pedimos los datos por consola
filas = int(input("¿Cuántas filas? "))
columnas = int(input("¿Cuántas columnas? "))
minimo = int(input("Valor mínimo (x): "))
maximo = int(input("Valor máximo (y): "))


mi_matriz = crear_matriz(filas, columnas, minimo, maximo)

for fila in mi_matriz:
    print(fila)
```

```python
# 2. Extrae:  
#    - La **diagonal principal**.
#    - La **diagonal secundaria**.

diagonal_principal = np.diag(mi_matriz)
diagonal_secundaria = np.diag(np.fliplr(mi_matriz))


print("\nDiagonal principal:", diagonal_principal)
print("Diagonal secundaria:", diagonal_secundaria)
```

```python
### 📌 📝 **Ejercicios (Broadcasting con filas)**

# 1. Crea una matriz 6x4 de enteros aleatorios entre 0 y 20.  
# 2. Calcula la media de cada fila.  
# 3. Resta a cada elemento de la matriz la media de su fila usando **broadcasting** (sin bucles).  
# 4. Explica:
#    - La forma de cada array antes de la operación.  
#    - Cómo NumPy aplica las reglas de broadcasting.


# 1 Crea una matriz 6x4 de enteros aleatorios entre 0 y 20.
matriz = np.random.randint(0,21,size=(6,4))
print("Matriz original 6x4\n" ,matriz)
print("\nTamaño\n" ,matriz.shape)

# 2. Calcula la media de cada fila. 
media_matriz = matriz.mean(axis=1,keepdims=True)
print("\nMedia por fila de la matriz\n" ,media_matriz)

 # 3. Resta a cada elemento de la matriz la media de su fila usando **broadcasting** (sin bucles).  
 
resultado = matriz - media_matriz
print("\nResultado:")
print(resultado)
 
 #4. Explica: 
 ## - La forma de cada array antes de la operación.  
    # comparando la forma de la matriz	(6, 4) vs la forma del array en este caso llamado medias (6, 1)
## - Cómo NumPy aplica las reglas de broadcasting?.
    #  Comparando las formas de derecha a izquierda, dimension por dimension, si son compatibles, es decir, son iguales o una de ellas es 1, entonces se estira para igualar a la otra.
    # Realizando una copia del array en memoria de manera "virtual", siendo más eficiente porque se ejecuta en memoria
```

```python
# *Objetivo:** poner en práctica todo lo visto de **NumPy** en una mini–pipeline numérica: creación y manipulación de arrays, *dtypes*, *broadcasting*, *slicing*, máscaras booleanas, operaciones vectorizadas, y bloque de **álgebra lineal** (normas, proyecciones, covarianza, autovectores) con una aplicación tipo “nube de puntos + proyección a una recta”.

# ---

# ### Contexto
# Tienes un conjunto de puntos 2D que representan **mediciones** de dos sensores \(`x`, `y`\). Quieres:
# 1) limpiar y normalizar,  
# 2) analizar distancias y outliers,  
# 3) proyectar sobre una dirección “de interés”,  
# 4) estudiar la covarianza y la dirección principal de variación (autovectores),  
# 5) comparar enfoques **vectorizados** vs. con bucles y medir tiempos.
import numpy as np
import matplotlib.pyplot as plt 
rng = np.random.default_rng(123)

# 1) Núcleo principal: nube alargada en torno a una recta y = 0.7x + ruido
n = 1200
x = rng.normal(0, 3, n)
y = 0.7 * x + rng.normal(0, 1.2, n)

# 2) Inyectamos algunos outliers
m = 30
xo = rng.normal(10, 1.5, m)
yo = rng.normal(-10, 1.5, m)

# 3) Unimos y formamos matriz de datos (n+m, 2)
X = np.column_stack([np.concatenate([x, xo]), np.concatenate([y, yo])]).astype(np.float64)

# Barajamos filas
rng.shuffle(X)

print("Forma de X:", X.shape)
print("Primeras filas:\n", X[:5])

# graficamos
plt.figure(figsize=(7, 6))                 # crea una figura nueva de 7x6 pulgadas
plt.scatter(X[:, 0], X[:, 1], s=10, alpha=0.5) 

plt.title("Conjunto de datos 2D")         
plt.xlabel("x (sensor 1)")                
plt.ylabel("y (sensor 2)")                
plt.grid(True, alpha=0.3)                 
plt.axis("equal")                         
plt.show()                                

```

```python
# PARTE A
# 1. Analiza el array de datos: muestra su tipo (`dtype`), forma (`shape`), número de dimensiones (`ndim`) y memoria ocupada (`nbytes`).  
print("Información del array\n") 
print("dtype:",X.dtype)
print("shape :", X.shape)               # forma: (filas, columnas) -> (1230, 2)
print("ndim  :", X.ndim)                # número de dimensiones (2: es una matriz)
print("nbytes:", X.nbytes, "bytes") 
```

```python
# 2) Compara el tamaño en memoria al convertirlo a `float32`. 
x_float32 = X.astype(np.float32)

print("nbytes float64:", X.nbytes, "bytes")
print("nbytes float64:", x_float32, "bytes")
print("Ahorro:", X.nbytes / x_float32.nbytes, "veces menos memoria")
```

```python
# 3. Extrae las primeras y últimas 10 filas.  

print("Primeras y ultimas 10 filas: \n")
primeras10 = X[:10]
ultimas10 = X[-10:]

print("Primeras 10:\n", primeras10)
print("Últimas 10:\n", ultimas10)
```

```python
# 4. Separa la columna `x` y la columna `y` en vectores independientes. 

col_x = X[:, 0]  # ":" = todas las filas, "0" = columna 0 (x)
col_y = X[:, 1]  # todas las filas, columna 1 (y)

print(" Columnas separadas\n")
print("col_x:", col_x.shape, "| col_y:", col_y.shape)   # ambos (1230,)
print("colx", col_x)
col_x[3]=7.77
print(col_x)
```

```python
# 5. Detecta posibles *outliers* con una máscara booleana: puntos cuya `x` o `y` tengan valor absoluto mayor que 6. 
mask_out = (np.abs(col_x) > 6) | (np.abs(col_y) > 6) 
print("\n--- 5) Outliers ---")
print("Tipo de la máscara:", mask_out.dtype)            # bool
print("Nº de puntos marcados:", mask_out.sum())
outliers = X[mask_out] 
print("Primeros outliers:\n", outliers[:5])
```

```python
### Parte B — Normalización y operaciones vectorizadas

# 6. Normaliza cada columna al rango `[0,1]` mediante **min–max scaling**.  
# 7. Estandariza los datos (media 0, desviación típica 1) por columnas.  
# 8. Calcula la distancia euclidiana de cada punto al centro de la nube.  
# 9. Identifica qué puntos están por encima del percentil 95 de las distancias.  

x_min = X.min(axis=0)                    # axis=0 -> opera "hacia abajo" (por columnas): array de 2 valores
x_max = X.max(axis=0)                    # máximo de cada columna
X_minmax = (X - x_min) / (x_max - x_min) # BROADCASTING: (1230,2) - (2,) se estira el (2,) a cada fila
print("Min-max\n")
print("Mínimos tras escalar:", X_minmax.min(axis=0))   # debe ser [0, 0]
print("Máximos tras escalar:", X_minmax.max(axis=0))   # debe ser [1, 1]
```

```python
# 7. Estandariza los datos (media 0, desviación típica 1) por columnas.  
media = X.mean(axis=0)
desv = X.std(axis=0)
print("media\n",media)
print("desviación\n",desv)
```

```python
#POC
data = np.array([
                    [20,1000],
                    [30,3000],
                    [40,5000],
                    [50,7000],
                    [60,9000]
                ])
print("Data:\n", data)

x_min=data.min(axis=0)
x_max = data.max(axis=0)

x_normalizados = (data-x_min) / (x_max-x_min)


print("\nMinimos\n",x_min)
print("\nMaximos\n",x_max)
print("\ndatos normalizados\n", x_normalizados)
```

```python
#8. Calcula la distancia euclidiana de cada punto al centro de la nube.
centro = X.mean(axis=0)                  # punto medio de la nube (2 valores)
dist_centro = np.linalg.norm(X - centro, axis=1)

print("\nDistancias al centro\n")
print("Centro:", centro)
print("Forma:", dist_centro.shape, "| min:", dist_centro.min(), "| max:", dist_centro.max())
```

```python
# 9. Identifica qué puntos están por encima del percentil 95 de las distancias.  
p95 = np.percentile(dist_centro, 95)     
mask_p95 = dist_centro > p95             

print("\n--- 9) Percentil 95 ---")
print("Umbral p95:", p95)
print("Puntos por encima:", mask_p95.sum())     # ~5 % de 1230 ≈ 61
print("Primeros puntos lejanos:\n", X[mask_p95][:5])
```
