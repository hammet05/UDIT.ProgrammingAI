# Reto integrador - Datase de ventas online
- En este ejercicio vas a trabajar con un **dataset simulado de ventas de una tienda online**.  
- Contiene 500 registros de pedidos con información de clientes, productos y precios.
#### **Descripción del dataset**

El archivo `ventas.csv` contiene las siguientes columnas:

- **order_id** → Identificador único del pedido.  
- **customer_id** → Identificador del cliente.  
- **product** → Producto comprado.  
- **category** → Categoría del producto (`Electrónica`, `Ropa`, `Hogar`, `Juguetes`).  
- **quantity** → Cantidad de unidades compradas.  
- **price** → Precio unitario del producto.  
- **discount** → Descuento aplicado al pedido (0–0.5).  
- **order_date** → Fecha del pedido.  
- **region** → Región de envío (`Norte`, `Sur`, `Este`, `Oeste`).  
- **delivered** → Indica si el pedido fue entregado (`Yes`/`No`).

El archivo `clientes.csv` contiene las siguientes columnas:

- **customer_id** → Identificador del cliente. 
- **age** → Edad del cliente.  
- **gender** → Género del cliente.

```python
#Conversión de tipos y optimización
# 1. Convierte `region` y `category` en **categorías**.  

# Memoria ANTES
mem_antes = df_ventas.memory_usage(deep=True).sum()

df_ventas["region"] = df_ventas["region"].astype("category")
df_ventas["category"] = df_ventas["category"].astype("category")



# 2. Convierte `delivered` en booleano.
df_ventas["delivered"] = df_ventas["delivered"].astype(bool)

# 3. Optimiza `quantity` como `int8`. 
df_ventas["quantity"] = df_ventas["quantity"].astype("int8")

# 4. Mide el **uso de memoria** antes y después de los cambios.  

# Memoria DESPUÉS
mem_despues = df_ventas.memory_usage(deep=True).sum()

print(f"Antes:   {mem_antes / 1024:.2f} KB")
print(f"Después: {mem_despues / 1024:.2f} KB")
print(f"Ahorro:  {(1 - mem_despues / mem_antes) * 100:.1f}%")


tabla_region = dict(enumerate(df_ventas["region"].cat.categories))
tabla_categorias = dict(enumerate(df_ventas["category"].cat.categories))
print("\ntablas de la relación category para region\n",tabla_region)
print("\ntablas de la relación category para categorias\n",tabla_categorias)
```

```python
# Filtrado e indexación
# 1. Obtén todos los pedidos con `quantity > 3` y `discount > 0.2`.  

filtro1 = df_ventas[(df_ventas["quantity"] > 3) & (df_ventas["discount"] > 0.2)]

# 2. Selecciona las columnas `product`, `quantity` y `discount` de las **primeras 5 filas**.  

filtro2 = df_ventas.loc[:4, ["product", "quantity", "discount"]] #loc usa etiquetas y el final es inclusivo: loc[:4] devuelve las filas con etiqueta 0, 1, 2, 3 y 4 (5 filas).
filtro_iloc = df_ventas[["product", "quantity", "discount"]].head(5) #iloc usa posiciones y el final es exclusivo: iloc[:5] también da 5 filas.

# 3. Selecciona con `.loc[]` todos los pedidos de la categoría `"Electrónica"` en la región `"Norte"`. 

filtro3 = df_ventas.loc[
    (df_ventas["category"] == "Electrónica") & (df_ventas["region"] == "Norte")
]

display(filtro1)
display(filtro2)
display(filtro_iloc)
display(filtro3)
```

```python
# Agregaciones y estadísticas
# 1. Calcula el **precio medio** por categoría de producto.  

categoria_precio_medio = df_ventas.groupby("category",observed=True)["price"].mean()
print("\nPrecio Medio de Categoria\n",categoria_precio_medio)

# 2. Calcula la **cantidad total vendida** por región.  
total_cantidad_vendida = df_ventas.groupby("category",observed=True)["quantity"].sum()
print("\ntotal_cantidad_vendida\n",total_cantidad_vendida)


# 3. Obtén la tasa de entrega (`delivered`) por categoría de producto.  
tasa_entrega = df_ventas.groupby("category", observed=True)["delivered"].mean()
print("\nTasa de entrega efectiva\n",tasa_entrega)
```

```python
# Combinación de DataFrames

# Leo clientes
df_clientes = pd.read_csv("../datos_ejercicio_integrador/clientes.csv")

print("\nVer top 3 clientes\n",df_clientes.head(3))
display("Ver top 3 ventas",df_ventas.head(3))
#categorizo gender

df_clientes["gender"] = df_clientes["gender"].astype("category")

tabla_categorias_gender = dict(enumerate(df_clientes["gender"].cat.categories))
print("\ntablas de la relación category para gender\n",tabla_categorias_gender)

# 1. Haz un `merge()` entre `ventas` y `clientes` para unir la información.  
merge_ventas_clientes = pd.merge(df_ventas,df_clientes,on="customer_id", how="left", validate="m:1" ) #filas repetidas 
display("Ver ventas merge con clientes",merge_ventas_clientes)

print("\nFilas deben coincidir con LEFT\n",df_ventas.shape, df_clientes.shape)   # las filas deben coincidir con left
print("\nPedidos sin cliente asociado\n", df_clientes.isna().sum())             # nulos = pedidos sin cliente asociado

# 2. Usa `concat()` para simular añadir **50 pedidos nuevos** al DataFrame original.  
nuevos_pedidos = df_ventas.sample(50, random_state=42)   # simulación: 50 filas de ejemplo
df_ampliado = pd.concat([df_ventas, nuevos_pedidos], ignore_index=True)
print("\nVentas ampliadas: + 50",df_ampliado.shape)   # debería tener 50 filas más que df_ventas
```

```python
# Vectorización vs. apply

# 1. Calcula una nueva columna `total` = `(quantity * price) * (1 - discount)` de dos formas:  
#    - Vectorizada.  
#    - Con `.apply()`.  

df_ventas_vect = df_ventas["total_vec"] = (df_ventas["quantity"] * df_ventas["price"]) * (1 - df_ventas["discount"])

print("Ventas Vectorizadas\n", df_ventas_vect);

# Con apply (fila a fila)
df_ventas_apply = df_ventas["total_apply"] = df_ventas.apply(lambda fila: (fila["quantity"] * fila["price"]) * (1 - fila["discount"]),  axis=1)
print("\nVentas con Aply\n", df_ventas_apply);


# 2. Usa `%timeit` para comparar ambas soluciones. 
print("\ncomparación:\t",np.allclose(df_ventas["total_vec"], df_ventas["total_apply"]))
```

```python
# Preparación para IA

# 1. Selecciona como features (`X`) las columnas: `quantity`, `price`, `discount`, `region`, `category`, `gender`.  

cols_X = ["quantity", "price", "discount", "region", "category", "gender"]
X = merge_ventas_clientes[cols_X].copy()

#    El target (`y`) será la columna `delivered`.  
y = merge_ventas_clientes["delivered"]
print("\nVentas Clientes Cols Seleccioandas\n", X, "\nColumna Delivered\n", y)

# 2. Aplica **One-Hot Encoding** a `region`, `category` y `gender`.  
X = pd.get_dummies(X, columns=["region", "category", "gender"], dtype=int)


# 3. Escala las columnas numéricas (`quantity`, `price`, `discount`) con **MinMaxScaler**.  
num_cols = ["quantity", "price", "discount"]
scaler = MinMaxScaler()

X[num_cols] = scaler.fit_transform(X[num_cols])
xp = X[num_cols]
print("\nColumas Númericas\n",num_cols)
print("\nColumas Escaladas\n",xp)

# 4. Comprueba que el DataFrame final contiene únicamente **valores numéricos** listos para un modelo de Machine Learning. 

print(X.dtypes)
print(X.select_dtypes(exclude="number").columns.tolist())   # debe salir []
print(X.isna().sum().sum())                                  # debe ser 0
display(X.head()) 
```
