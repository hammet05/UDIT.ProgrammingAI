```python
# 3. Rellena los valores nulos en `region` con la **moda** de la columna.
region_nulos = df_ventas["region"].isna().sum()
display("Total valores nulos en columna región\n",region_nulos)

frecuencia_region = df_ventas["region"].value_counts()
display("Frecuencia valores nulos en columna región\n",frecuencia_region)

moda_region = df_ventas["region"].mode()[0]
print("Frecuencia de region nula\n",moda_region)

df_ventas["region"] = df_ventas["region"].fillna(moda_region)
df_ventas
```

# KAGGLE
1. Descargas un dataset en CSV desde Kaggle.
2. Lo cargas con pd.read_csv().
3. Tras limpieza y transformación, lo guardas en Parquet para acelerar futuros experimentos.
4. Exportas resultados intermedios en Excel para mostrarlos a un analista de negocio.
5. Subes los resultados a una base de datos SQL de producción.


```python
# 1. CSV

print(os.getcwd())            # directorio de trabajo actual
print(os.listdir())           # archivos que hay en él
print(os.path.exists("../heart_disease_prediction_2026.csv"))   # ¿existe el archivo?

df = pd.read_csv("../heart_disease_prediction_2026.csv",
    sep=",",               # separador (por defecto ",")
    encoding="utf-8",      # o "latin-1" si hay problemas con tildes
    # usecols=["student_id", "study_hours","exam_score"],  # leer solo algunas columnas
    nrows=1000,            # leer solo las primeras 1000 filas        
)

df
```

```python
# 3. Tras limpieza y transformación
#df = df.drop_duplicates()
print("Filas:", len(df))
print("Duplicados:", df.duplicated().sum())

cols_cat = ["sex", "region", "income_level", "smoking_status",
            "alcohol_intake", "physical_activity", "resting_ecg", "st_slope"]
df[cols_cat] = df[cols_cat].astype("category")

# Columnas 0/1 como booleanas (opcional, más legible)
cols_bin = ["diabetes", "hypertension", "family_history", "previous_heart_event",
            "fasting_blood_sugar", "exercise_angina" ]
df[cols_bin] = df[cols_bin].astype(bool)
df["heart_disease"] = df["heart_disease"].astype(int)

df
```

```python
# Guardas en Parquet

df.to_parquet("heart_disease_limpio.parquet", index=False)

# Próximas veces
df = pd.read_parquet("heart_disease_limpio.parquet")

df
```

```python
# 4. Exportas resultados intermedios en Excel para mostrarlos a un analista de negocio.
resumen_sexo = df.groupby("sex", observed=True)["heart_disease"].mean().mul(100).round(1).reset_index(name="% con enfermedad")
resumen_fumador = df.groupby("smoking_status", observed=True)["heart_disease"].mean().mul(100).round(1).reset_index(name="% con enfermedad")

with pd.ExcelWriter("informe_corazon.xlsx") as writer:
    resumen_sexo.to_excel(writer, sheet_name="Por sexo", index=False)
    resumen_fumador.to_excel(writer, sheet_name="Por tabaquismo", index=False)
    df.head(1000).to_excel(writer, sheet_name="Muestra", index=False)
```
